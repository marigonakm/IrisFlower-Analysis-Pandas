```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

```python
plt.plot(df['PetalLengthCm'], df['PetalWidthCm'], 'o', color='purple')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.title('Petal Length vs Petal Width')
plt.show()
```

```python
sns.regplot(x='PetalLengthCm', y='PetalWidthCm', data=df, color='purple')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.title('Petal Length vs Petal Width')
```

```python
sns.regplot(x='PetalLengthCm', y='SepalWidthCm', data=df, color='purple')
plt.title("Relationship between Petal Length and Sepal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.grid(True)
plt.show()
```

```python
sns.lineplot(x='PetalLengthCm', y='SepalWidthCm', data=df, errorbar=None, color='pink')
plt.title("Petal Length vs Sepal Width")
plt.show()

```

```python
df.groupby('Species')[['PetalLengthCm', 'SepalWidthCm']].mean()
```
